```python
from huggingface_hub import login
import os

from dotenv import load_dotenv
from transformers import AutoModelForCausalLM, AutoTokenizer
```

```python
load_dotenv()
login(os.getenv("HF_TOKEN"), add_to_git_credential=True)

model = AutoModelForCausalLM.from_pretrained("meta-llama/Llama-3.2-1B", device_map="auto")
model
```

```python
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-1B")
tokenizer
```

```python
encoded_ids = tokenizer.encode("Hello, my dog is cute")
encoded_ids
```

```python
decoded = tokenizer.decode(encoded_ids)
decoded
```

```python
tokenizer.tokenize("Hello, my dog is cute")
```
